```python
student = {
    "Name": "Laxmikanth",
    "Roll No": 2520030018,
    "Branch": "CSE",
    "CGPA": 9.43
}

print("Student Details:")

for key, value in student.items():
    print(key, ":", value)
```

```python
book = {
    "Title": "Python Programming",
    "Author": "John",
    "Price": 550,
    "Publisher": "ABC Publications"
}

print("Book Details:")

for key, value in book.items():
    print(key, "=", value)
```

Program 6: Conditional Statements
Write a Python program to classify a student's CGPA into High, Mid, or Low categories using if-elif-else.
Write a program to check whether missing values are present in a dataset using conditional statements.
Write a program to determine whether a student is eligible for placement based on CGPA.
Write a program that uses if-elif-else to assign grades based on marks.


```python
cgpa = float(input("Enter CGPA: "))

if cgpa >= 8.5:
    print("Category: High")
elif cgpa >= 6.5:
    print("Category: Mid")
else:
    print("Category: Low")
```

```python
dataset = [45, 67, None, 89, 72]

if None in dataset:
    print("Missing values are present in the dataset.")
else:
    print("No missing values found.")
```

```python
cgpa = float(input("Enter CGPA: "))

if cgpa >= 7.0:
    print("Student is Eligible for Placement.")
else:
    print("Student is Not Eligible for Placement.")
```

```python
marks = int(input("Enter Marks: "))

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 40:
    print("Grade: D")
else:
    print("Grade: F")
```

Program 7: Loops
Write a Python program to display all elements in a list using a for loop.
Write a program to calculate the average of numbers stored in a list.
Write a program to display only the values greater than a given threshold using a loop.
Write a program to find the maximum and minimum values in a list using loops.
Write a program to count the number of elements satisfying a given condition using a loop.

```python
numbers = [12, 25, 37, 48, 59]

print("Elements in the list:")

for num in numbers:
    print(num)
```

```python
marks = [80, 75, 90, 85, 70]

total = 0

for mark in marks:
    total += mark

average = total / len(marks)

print("Marks:", marks)
print("Average =", average)
```

```python
numbers = [10, 45, 67, 23, 89, 34]
threshold = 40

print("Values greater than", threshold, ":")

for num in numbers:
    if num > threshold:
        print(num)
```

```python
numbers = [25, 14, 68, 9, 41, 55]

maximum = numbers[0]
minimum = numbers[0]

for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num

print("Maximum Value =", maximum)
print("Minimum Value =", minimum)
```

```python
numbers = [12, 15, 18, 21, 24, 30, 35]

count = 0

for num in numbers:
    if num % 2 == 0:
        count += 1

print("List:", numbers)
print("Number of Even Elements =", count)
```
